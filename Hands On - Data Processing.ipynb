## Melihat jumlah orang per provinsi

```python
cia_dataset.groupby('province')['registration_id'].count()
```

## Melihat rata-rata score tiap kelas

```python
cia_dataset.groupby('class')['score'].mean().reset_index()
```

```python
cia_dataset.groupby('class')['score'].mean().reset_index().sort_values('score',ascending=False)
```
